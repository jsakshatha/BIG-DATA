# PySpark RDD: Transformations and Actions



**Transformations (lazy):** `map()`, `filter()`, `sortBy()`

**Actions (trigger execution):** `collect()`, `take()`, `first()`, `count()`, `reduce()`

> Important: A transformation creates a new RDD and records the operation. An action triggers execution of the required transformations.


## 1. Start Spark and create a Python list




```python
# STEP 1: Install PySpark in Colab
# ---------------------------------------------------------
!pip install pyspark -q   # -q keeps the install output quiet
```

```python

# ---------------------------------------------------------
# STEP 2: Import the SparkSession class
# ---------------------------------------------------------
from pyspark.sql import SparkSession

# ---------------------------------------------------------
# STEP 3: Create the SparkSession (the modern entry point)
# appName      -> name shown in the Spark UI
# master       -> "local[2]" tells Spark: run locally using 2 worker threads (2 CPUs)
# getOrCreate  -> reuse the existing session if one is already running

# ---------------------------------------------------------
spark = (SparkSession.builder
         .appName("RDD_Basic")
         .master("local[1]")
         .getOrCreate())

# ---------------------------------------------------------
# STEP 4: Get the SparkContext from the session

# ---------------------------------------------------------
sc = spark.sparkContext

# Print confirmation
print("SparkSession created successfully!")
print("Spark version:", spark.version)
print("Running with:", sc.master)   # Should print local[2]
print("App name:", sc.appName)
```

```python
# from pyspark import SparkContext

# # Create SparkContext (run this cell once)
# sc = SparkContext.getOrCreate()

# Marks of 10 students
marks = [10, 20, 30, 40, 50, 20, 60, 70, 80, 90]

# Convert the Python list into an RDD with 2 partitions
rdd = sc.parallelize(marks)

print("Original Python list:", marks)
print("Number of partitions:", rdd.getNumPartitions())
```

```python
# Show the records in each partition
print(rdd.glom().collect())
```

## 2. Transformation: `map()`



```python
# Double every mark
rdd_map = rdd.map(lambda mark: mark * 2)

print("Transformation created.")

# This action triggers execution
print("Actual records:", rdd_map.collect())
```

## 3. Transformation: `filter()`


`filter()` keeps only the elements that satisfy the condition.


```python
# Keep marks greater than 50
rdd_filter = rdd.filter(lambda mark: mark > 50)

# collect() is an action and triggers the filter
print(rdd_filter.collect())
```

## 4. Transformation: `sortBy()`



`sortBy()` creates a new RDD with the records sorted by the key or value returned by the function.


```python
# Sort marks in ascending order
rdd_sorted = rdd.sortBy(lambda mark: mark)

print("Ascending:", rdd_sorted.collect())

# Sort marks in descending order
rdd_sorted_desc = rdd.sortBy(lambda mark: mark, ascending=False)

print("Descending:", rdd_sorted_desc.collect())
```

```python
result = rdd.collect()
print(result)
```

```python
print(rdd.take(3))
```

```python
print(rdd.first())
```

### Action 4: `count()`



`count()` returns the number of elements in the RDD.


```python
print("Number of records:", rdd.count())
```

### Action 5: `reduce()`



`reduce()` combines the elements using the function supplied. Here, we add the marks together.


```python
total_marks = rdd.reduce(lambda a, b: a + b)
print("Total marks:", total_marks)
```

## 5. Stop Spark when finished



```python
sc.stop()
```
